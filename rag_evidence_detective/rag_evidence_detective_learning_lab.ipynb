## Look at the raw text

First explore the document; note headings and causal facts separated across paragraphs.

```python
from pathlib import Path
doc=Path("rag_evidence_detective/sample_documents.txt").read_text()
print(doc)
```

## Split into paragraphs

We deliberately use a simple split, so that we can see its failure later.

```python
from rag_evidence_detective.logic import chunk_paragraphs
chunks=chunk_paragraphs({"sample_report":doc})
for c in chunks:print(c.id,len(c.text),repr(c.text[:90]))
```

## Score retrieval manually

Our baseline is token-set overlap, not a vector database. Count shared terms in Python.

```python
from rag_evidence_detective.logic import words,retrieve
question="What caused the aerospace revenue decline?"
for c in chunks:print(c.id,len(words(question)&words(c.text)))
matches=retrieve(question,chunks,k=2)
[(p.id,score) for p,score in matches]
```

## Build final model context

The complete context should be inspectable and contain source IDs.

```python
from rag_evidence_detective.logic import build_context,answer_prompt
context=build_context(matches)
print(answer_prompt(question,context))
```

## Deliberately break retrieval

Paraphrase the question with words not present in the evidence. Compare scores; do not invent a retrieval success metric.

```python
for q in ["What caused the aerospace revenue decline?","Why did aircraft clients drop their deals?"]:
    print(q,[(p.id,score) for p,score in retrieve(q,chunks,2)])
```
