## Inspect the input assumptions

Our example uses synthetic millions of USD. Every percentage has a clear denominator.

```python
from finance_scenario_simulator.logic import simulate
base=740; growth=7; cancellations=8; cost=70
print(base,growth,cancellations,cost)
```

## Compute without an LLM

We calculate planned revenue, cancellations, expenses, and profit from first principles.

```python
result=simulate(base,growth,cancellations,cost)
result
```

## Visualize preprocessing

Before trusting any generated interpretation, make the calculation steps visible.

```python
import matplotlib.pyplot as plt
labels=["Base","Planned","After cancellations","Operating profit"]
values=[result["base_revenue_m"],result["planned_revenue_m"],result["revenue_after_cancellations_m"],result["operating_profit_m"]]
plt.figure(figsize=(9,3.5));plt.bar(labels,values);plt.xticks(rotation=15);plt.ylabel("Million USD");plt.title("Synthetic revenue scenario — calculated in Python");plt.tight_layout();plt.show()
```

## Explain with LLM (optional)

```python
from finance_scenario_simulator.logic import explanation_prompt
print(explanation_prompt(result))
# from ibm_client import generate
# print(generate(explanation_prompt(result)))
```

## Break the assumptions

Try extreme cancellations or bad input. We want to learn the boundary between a coherent toy model and actual business forecasting.

```python
for cancellation in [0,20,80,100]:
    print(cancellation,simulate(100,10,cancellation,70))
```
